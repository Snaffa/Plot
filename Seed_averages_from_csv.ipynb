# Seed averages from csv
Current slope and avalanche buildup time vs overvoltage, read from the csv files saved by
the seed analyses. The rows of each csv are in order of decreasing overvoltage
(row 0 is the highest, the last row is 1 V).

Figures are saved in `SAVE_DIR` as `.png`, `.pdf` and editable `.mplfig`
(reopen with `python open_fig.py <file>.mplfig` or `mcplot.load_figure`).

```python
# Interactive figures (zoom/pan) need ipympl; otherwise fall back to static figures
try:
    import ipympl  # noqa: F401
    get_ipython().run_line_magic("matplotlib", "widget")
except ImportError:
    get_ipython().run_line_magic("matplotlib", "inline")
import os

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from mcplot import save_figure, set_style

set_style(35)

SAVE_FIGURES = True  # save every figure as .png, .pdf and editable .mplfig
SAVE_DIR = "/home/angelo/scripts/plot/NUSOD2026/active_region_study"


def save(fig, name):
    """Save fig as SAVE_DIR/name.{png,pdf,mplfig} (reopen the .mplfig with open_fig.py)."""
    if SAVE_FIGURES:
        save_figure(fig, os.path.join(SAVE_DIR, name), formats=("png", "pdf"), dpi=300, bbox_inches="tight")
```

```python
def plot_csv_vs_overvoltage(csv_files, ylabel, legend_loc, column=1, scale=1.0, std_column=None):
    """One curve per csv file ({label: path}): column `column` (times `scale`) vs overvoltage.

    With `std_column`, that column (times `scale`) is drawn as error bars.
    """
    fig, ax = plt.subplots(figsize=(15, 10))
    colors = plt.cm.tab10.colors
    n_max = 0
    for idx, (label, path) in enumerate(csv_files.items()):
        df = pd.read_csv(path)
        y = df.iloc[:, column].to_numpy() * scale
        vov = np.arange(len(y), 0, -1, dtype=float)
        n_max = max(n_max, len(y))
        color = colors[idx % len(colors)]
        if std_column is None:
            ax.plot(vov, y, "o--", markersize=10, color=color, label=label)
        else:
            ax.errorbar(vov, y, yerr=df.iloc[:, std_column].to_numpy() * scale, fmt="o--", markersize=10,
                        capsize=5, capthick=2, elinewidth=2, color=color, label=label)
    ax.set_xticks(np.arange(1, n_max + 1))
    ax.set_xlabel("Overvoltage, V", fontsize=40)
    ax.set_ylabel(ylabel, fontsize=40)
    ax.grid(True, linestyle="--", alpha=0.7)
    ax.legend(ncol=int(np.ceil(len(csv_files) / 4)), loc=legend_loc)
    fig.tight_layout()
    return fig
```

# Current slope

```python
csv_dir = "/home/angelo/scripts/plot/NUSOD2026/active_region_study/current_slope"
csv_files = {  # label: csv file
    "1.0 µm": f"{csv_dir}/1.0 µm.csv",
    "1.1 µm": f"{csv_dir}/1.1 µm.csv",
    "1.2 µm": f"{csv_dir}/1.2 µm.csv",
    "1.3 µm": f"{csv_dir}/1.3 µm.csv",
    # "1.4 µm": f"{csv_dir}/1.4 µm.csv",
    # "1.5 µm": f"{csv_dir}/1.5 µm.csv",
    # "1.6 µm": f"{csv_dir}/1.6 µm.csv",
    # "1.7 µm": f"{csv_dir}/1.7 µm.csv",
}
```

```python
plt.close("all")
fig = plot_csv_vs_overvoltage(csv_files, "Current growth rate, A/ps", "upper left",
                              column=1, std_column=2, scale=1e-12)
save(fig, "current_slope/slope_vs_overvoltage")
plt.show()
```
